```python
import numpy as np
import pandas as pd
import os

PROCESSED_DIR = os.path.join("data", "processed")
if not os.path.exists(PROCESSED_DIR):
    root = os.path.abspath(os.path.join(os.getcwd(), ".."))
    cand = os.path.join(root, PROCESSED_DIR)
    if os.path.exists(cand):
        PROCESSED_DIR = cand
        os.chdir(root)
        print(f"项目根目录: {root}")

for split in ['train', 'val', 'test']:
    sig_path = os.path.join(PROCESSED_DIR, split, 'signals.npy')
    lbl_path = os.path.join(PROCESSED_DIR, split, 'labels.csv')
    
    signals = np.load(sig_path)
    labels_df = pd.read_csv(lbl_path)
    
    print(f"\n=== {split} ===")
    print(f"  signals.npy 形状: {signals.shape} (dtype={signals.dtype})")
    print(f"  labels.csv  行数: {len(labels_df)}")
    print(f"  列名: {list(labels_df.columns)}")
    
    # 如果 signals 是 3D (样本, 导联, 时间点)，显示统计信息
    if signals.ndim == 3:
        print(f"  信号范围: [{signals.min():.4f}, {signals.max():.4f}]")
        print(f"  信号均值: {signals.mean():.4f}, 标准差: {signals.std():.4f}")

# 同时检查多标签文件
label_dir = os.path.join(PROCESSED_DIR, 'labels')
if os.path.exists(label_dir):
    print("\n=== 多标签文件 ===")
    for f in os.listdir(label_dir):
        if f.endswith('.npy'):
            arr = np.load(os.path.join(label_dir, f))
            print(f"  {f}: 形状 {arr.shape}")
```
