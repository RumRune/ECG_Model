# PTB‑XL 数据探索

```python
# ============================
# 01_data_exploration.py
# PTB‑XL 数据集初步探索（完整加载 + 修正版）
# ============================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# ---------- 全局设置 ----------
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 150)

# ==========================================
# 0. 加载数据（请根据实际路径修改）
# ==========================================
DATA_PATH = os.path.join("data", "raw", "ptb-xl_1.0.2", "ptbxl_database.csv")   # 相对项目根

# 若当前工作目录不在项目根，自动定位到项目根
if not os.path.exists(DATA_PATH):
    root = os.path.abspath(os.path.join(os.getcwd(), ".."))
    cand = os.path.join(root, DATA_PATH)
    if os.path.exists(cand):
        DATA_PATH = cand
        os.chdir(root)
        print(f"项目根目录: {root}")
# 如果默认路径仍不存在，手动让用户输入
if not os.path.exists(DATA_PATH):
    DATA_PATH = input("请输入 ptbxl_database.csv 的完整路径：").strip()

print(f"正在从 {DATA_PATH} 加载数据 ...")
db = pd.read_csv(DATA_PATH)
print("加载完成！\n")

# ==========================================
# 1. 数据基本概览
# ==========================================
print("=" * 50)
print(f"数据集尺寸: {db.shape}")
print("=" * 50)

print("\n【列数据类型】")
print(db.dtypes)

print("\n【前 5 行样本】")
display(db.head())

print("\n【全部列名】")
print(db.columns.tolist())

# ==========================================
# 2. 缺失值检测
# ==========================================
missing = db.isnull().sum()
missing_percent = (missing / len(db)) * 100
missing_df = pd.DataFrame({
    '缺失数量': missing,
    '百分比(%)': missing_percent
}).sort_values(by='缺失数量', ascending=False)

print("\n【缺失值统计（Top 20）】")
display(missing_df.head(20))

# ==========================================
# 3. 描述性统计
# ==========================================
print("\n【数值型字段描述】")
display(db.describe(include=[np.number]))

print("\n【分类型 / 布尔型字段描述】")
display(db.describe(include=['object', 'category', 'bool']))

# ==========================================
# 4. 核心标签分布（修正了列名）
# ==========================================
print("\n【Fold 分布（训练 / 验证 / 测试拆分）】")
fold_counts = db['strat_fold'].value_counts().sort_index()
print(fold_counts)
print(f"总计: {fold_counts.sum()} 条记录")

print("\n【人类专家验证情况（validated_by_human）】")
human_counts = db['validated_by_human'].value_counts()
print(human_counts)
print(f"人类验证比例: {db['validated_by_human'].mean() * 100:.2f}%")

print("\n【验证方式编码（validated_by）】")
print(db['validated_by'].value_counts().sort_index())

# ==========================================
# 5. 可视化：核心分布
# ==========================================
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 5.1 Fold 分布（柱状图）
ax = axes[0, 0]
fold_counts.plot(kind='bar', ax=ax, color='steelblue', edgecolor='black')
ax.set_title('Fold Distribution (10 stratified folds)', fontsize=13)
ax.set_xlabel('Fold Number')
ax.set_ylabel('Record Count')
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=9)

# 5.2 专家验证饼图（✅ 已修正：使用 'validated_by_human'）
ax = axes[0, 1]
db['validated_by_human'].value_counts().plot.pie(
    autopct='%1.1f%%',
    ax=ax,
    labels=['Human Validated', 'Not Human Validated'],
    colors=['#2ca02c', '#d62728'],
    startangle=90,
    textprops={'fontsize': 11}
)
ax.set_title('Expert Validation (Human)', fontsize=13)
ax.set_ylabel('')

# 5.3 年龄分布
ax = axes[1, 0]
if 'age' in db.columns:
    ax.hist(db['age'].dropna(), bins=30, color='darkorange', edgecolor='black', alpha=0.7)
    ax.set_title('Patient Age Distribution', fontsize=13)
    ax.set_xlabel('Age (years)')
    ax.set_ylabel('Frequency')
else:
    ax.text(0.5, 0.5, 'No "age" column', ha='center', va='center')
    ax.set_title('Age Distribution')

# 5.4 性别分布
ax = axes[1, 1]
if 'sex' in db.columns:
    sex_counts = db['sex'].value_counts()
    sex_labels = {0: 'Male', 1: 'Female'}
    sex_counts.index = sex_counts.index.map(sex_labels)
    sex_counts.plot(kind='bar', ax=ax, color=['skyblue', 'lightcoral'], edgecolor='black')
    ax.set_title('Sex Distribution', fontsize=13)
    ax.set_xlabel('Sex')
    ax.set_ylabel('Count')
    ax.set_xticklabels(sex_counts.index, rotation=0)
else:
    ax.text(0.5, 0.5, 'No "sex" column', ha='center', va='center')
    ax.set_title('Sex Distribution')

plt.tight_layout()
plt.show()

# ==========================================
# 6. 其他关键字段速查
# ==========================================
print("\n【信号噪声标签分布】")
noise_cols = ['baseline_drift', 'static_noise', 'burst_noise',
              'electrodes_problems', 'extra_beats', 'pacemaker']
for col in noise_cols:
    if col in db.columns:
        print(f"\n{col}:")
        print(db[col].value_counts(normalize=True).head())

print("\n【设备分布（Top 10）】")
if 'device' in db.columns:
    print(db['device'].value_counts().head(10))

print("\n【站点分布】")
if 'site' in db.columns:
    print(db['site'].value_counts())

print("\n===== 数据探索完成 =====")
```
